# Package comparisons

Here's a detailed comparison of the Python music packages `music21` and `mingus`:

| Feature                     | music21                                             | mingus                                                    |
|-----------------------------|-----------------------------------------------------|-----------------------------------------------------------|
| **Primary Focus**           | Computational musicology, music analysis, and music theory | Music theory, MIDI file manipulation, and playback support |
| **Core Components**         | Notes, chords, streams, musicXML, MIDI, LilyPond, and corpus search tools | Notes, chords, scales, keys, intervals, MIDI events, and containers (e.g., bars, tracks, compositions) |
| **MIDI Support**            | Yes, including conversion and playback              | Yes, including MIDI file creation and real-time playback using FluidSynth |
| **MusicXML Support**        | Yes, for reading and writing music notation         | No direct support                                         |
| **LilyPond Integration**    | Yes, for generating sheet music                     | Yes, for generating PNGs and PDFs of musical scores       |
| **Ease of Use**             | Comprehensive but can be complex for beginners      | More straightforward for basic music theory tasks         |
| **Educational Tools**       | Extensive documentation and user guide              | Tutorials available for various music theory concepts     |
| **Operating Systems**       | Cross-platform (Windows, macOS, Linux)              | Cross-platform (Windows, macOS, Linux)                    |
| **Licensing**               | BSD License                                         | BSD License                                               |
| **Community and Support**   | Active community with contributions and mailing list| Smaller community with GitHub support                     |
| **Latest Version**          | 9.1.0 (as of June 2023)                             | 0.5.1 (as of latest update)                               |
| **Data Formats**            | Supports multiple formats like MIDI, musicXML, and LilyPond | Primarily MIDI and LilyPond                               |
| **Special Features**        | Advanced music analysis tools, corpus searching, post-tonal tools, graphing and plotting | Focus on music theory and real-time MIDI playback         |
| **Documentation**           | Extensive with a detailed user guide, module reference, and developer guide | Good documentation with examples and tutorials            |

### Detailed Features

**music21**:
- **Data Handling**: music21 does not have a native data format but works well with many common formats, making it flexible for various music data manipulations.
- **Music Analysis**: Includes tools for advanced music analysis such as Roman numeral analysis, chord analysis, and corpus searching.
- **Educational Resources**: Offers a comprehensive user guide covering installation, basic to advanced usage, and extensions.
- **Graphing and Plotting**: Supports advanced graphing and plotting capabilities for visualizing musical data.

**mingus**:
- **Core Modules**: Includes modules for notes, intervals, chords, scales, and keys, making it a robust tool for music theory.
- **Containers**: Provides data structures to group notes (e.g., NoteContainers, Bars, Tracks, Compositions).
- **MIDI Functionality**: Can convert musical objects to MIDI events and supports real-time playback with FluidSynth.
- **LilyPond Integration**: Allows the creation of sheet music in PNG and PDF formats using LilyPond.
- **User-Friendly**: More straightforward for basic music theory tasks with practical examples and tutorials available.

Both packages serve different needs within the realm of music programming in Python. music21 is more suitable for those interested in deep musicological analysis and comprehensive music data handling, whereas mingus is tailored for users focusing on music theory and real-time MIDI manipulation.

For more information on `music21`, visit the [music21 documentation](https://web.mit.edu/music21). For details on `mingus`, refer to the [mingus documentation](http://bspaans.github.io/python-mingus/).

````python
from tonal.util import *
from tonal.counterpoint import *

def create_score_from_tracks(tracks: List[Stream]) -> Score:
    """
    Creates a music21 Score from a list of tracks (Stream objects).

    Args:
        tracks (List[Stream]): A list of Stream objects, each representing a track (voice).

    Returns:
        Score: A Score object containing the tracks as separate parts.

    Examples:
        >>> stream1 = Stream([Note('C4'), Note('E4'), Note('G4')])
        >>> stream2 = Stream([Note('A4'), Note('C5'), Note('E5')])
        >>> score = create_score_from_tracks([stream1, stream2])
        >>> len(score.parts)
        2
    """
    parts = [Part(iter(track)) for track in tracks]
    return Score(parts)

create_score_from_tracks(tracks).show('text')
````

````python
from music21 import note, scale, stream

def tile_motif_with_translations(motif, scale, translations):
    """
    Generate a sequence of notes by tiling translations of the given motif within the given scale.
    
    Parameters:
    motif (list): A list of music21.note.Note instances representing the motif.
    scale (music21.scale.AbstractScale): A music21 scale object within which the translations should stay.
    translations (list): A list of integers representing the offsets for the motif translations.
    
    Returns:
    music21.stream.Stream: A music21 Stream containing the resulting sequence of notes.
    """
    result = stream.Stream()
    
    # Convert the scale to a list of pitches
    scale_pitches = [p.nameWithOctave for p in scale.getPitches()]
    
    for trans in translations:
        for n in motif:
            # Calculate the index in the scale
            pitch_name = n.nameWithOctave
            index_in_scale = scale_pitches.index(pitch_name)
            new_index = (index_in_scale + trans) % len(scale_pitches)
            
            # If the translation is negative, adjust the octave if necessary
            if index_in_scale + trans < 0:
                new_pitch = note.Note(scale_pitches[new_index])
                new_pitch.octave -= 1
            else:
                new_pitch = note.Note(scale_pitches[new_index])
                
            result.append(new_pitch)
    
    return result
````

````python
from music21 import note, scale, stream

def tile_motif_with_translations(motif, scale, translations):
    """
    Generate a sequence of notes by tiling translations of the given motif within the given scale.
    
    Parameters:
    motif (list): A list of music21.note.Note instances representing the motif.
    scale (music21.scale.AbstractScale): A music21 scale object within which the translations should stay.
    translations (list): A list of integers representing the offsets for the motif translations.
    
    Returns:
    music21.stream.Stream: A music21 Stream containing the resulting sequence of notes.
    """
    result = stream.Stream()
    
    # Convert the scale to a list of pitches
    scale_pitches = [p.nameWithOctave for p in scale.getPitches()]
    
    for trans in translations:
        for n in motif:
            # Calculate the index in the scale
            pitch_name = n.nameWithOctave
            index_in_scale = scale_pitches.index(pitch_name)
            new_index = (index_in_scale + trans) % len(scale_pitches)
            
            # Calculate the octave change
            octave_change = (index_in_scale + trans) // len(scale_pitches)
            
            # Handle negative indices for octave change
            if (index_in_scale + trans) < 0:
                octave_change -= 1
            
            # Create the new pitch
            new_pitch = note.Note(scale_pitches[new_index])
            new_pitch.octave += octave_change
            
            result.append(new_pitch)
    
    return result

# Example usage
motif = [note.Note('E4'), note.Note('C4'), note.Note('B4'), note.Note('D4')]
s = scale.MajorScale('C4')
translations = [0, -2]

result_stream = tile_motif_with_translations(motif, s, translations)
result_stream.show('text')
````

````python
from music21 import note, scale, stream

def is_motif(item):
    """
    Check if the item is a motif (list of music21.note.Note instances) or a list of motifs.
    
    Parameters:
    item: The item to check.
    
    Returns:
    bool: True if the item is a motif, False if it is a list of motifs.
    """
    return all(isinstance(i, note.Note) for i in item)

def tile_motif_with_translations(motif, scale, translations):
    """
    Generate a sequence of notes by tiling translations of the given motif within the given scale.
    
    If the motif is multi-channel (a list of motifs), apply the translation recursively to each motif.
    
    Parameters:
    motif (list): A list of music21.note.Note instances or a list of motifs.
    scale (music21.scale.AbstractScale): A music21 scale object within which the translations should stay.
    translations (list): A list of integers representing the offsets for the motif translations.
    
    Returns:
    music21.stream.Score: A music21 Score containing the resulting sequence of notes.
    """
    result = stream.Score()
    
    if is_motif(motif):
        # Handle single motif
        part = stream.Part()
        scale_pitches = [p.nameWithOctave for p in scale.getPitches()]
        
        for trans in translations:
            for n in motif:
                # Calculate the index in the scale
                pitch_name = n.nameWithOctave
                index_in_scale = scale_pitches.index(pitch_name)
                new_index = (index_in_scale + trans) % len(scale_pitches)
                
                # If the translation is negative, adjust the octave if necessary
                if index_in_scale + trans < 0:
                    new_pitch = note.Note(scale_pitches[new_index])
                    new_pitch.octave -= 1
                else:
                    new_pitch = note.Note(scale_pitches[new_index])
                    
                part.append(new_pitch)
        
        result.append(part)
    
    else:
        # Handle multi-channel motif
        for voice in motif:
            voice_part = tile_motif_with_translations(voice, scale, translations)
            result.append(voice_part)
    
    return result

# Example usage
motif1 = [note.Note('E4'), note.Note('F4'), note.Note('G4')]
motif2 = [note.Note('C4'), note.Note('D4'), note.Note('E4')]
multi_voice_motif = [motif1, motif2]
s = scale.MajorScale('C4')
translations = [0, -1, 2]

result_score = tile_motif_with_translations(multi_voice_motif, s, translations)
result_score.show('text')
````

````python
import io
from mingus.containers import Note, Bar, Track, Composition
from mingus.core import scales
from mingus.midi import midi_file_out
from midi2audio import FluidSynth
from IPython.display import Audio

# Define the melody pattern
melody_pattern = [Note("C"), Note("E"), Note("G")]

# Define the scale
my_scale = scales.Major("C")

# Create a track to hold bars
track = Track()

# Function to transpose and add notes to the track
def add_transposed_pattern(pattern, scale, steps, track):
    for step in steps:
        bar = Bar()
        for n in pattern:
            transposed_note = Note(n)
            transposed_note.transpose(step)
            if transposed_note.name in scale.ascending():
                bar.place_notes(transposed_note, 4)
        track.add_bar(bar)

# Define transpositions (in semitones)
transpositions = [0, 2, 4, 5, 7, 9, 11]  # Example steps within C major scale
add_transposed_pattern(melody_pattern, my_scale, transpositions, track)

# Create a composition and add the track
composition = Composition()
composition.add_track(track)

# Write the track to a MIDI file
midi_io = io.BytesIO()
midi_file_out.write_Composition(midi_io, composition)

# Move the cursor to the beginning of the BytesIO object
midi_io.seek(0)

# Convert MIDI to audio using FluidSynth
fs = FluidSynth()
audio_io = io.BytesIO()
fs.midi_to_audio(midi_io, audio_io)

# Move the cursor to the beginning of the audio BytesIO object
audio_io.seek(0)

# Load and play the audio data from the BytesIO object
audio = Audio(audio_io.read(), rate=44100)
audio
````

````python
import mido
from mido import MidiFile, MidiTrack, Message

def generate_midi(chord_sequence, style_patterns, output_file):
    midi = MidiFile()
    track = MidiTrack()
    midi.tracks.append(track)

    track.append(Message('program_change', program=0, time=0))  # Acoustic Grand Piano

    for chord, duration in chord_sequence:
        pattern = style_patterns.get(chord, [])
        if pattern:
            # Add note-on events for all notes in the chord simultaneously
            for note in pattern:
                track.append(Message('note_on', note=note, velocity=64, time=0))
            # Add note-off events for all notes in the chord after the duration
            for note in pattern:
                track.append(Message('note_off', note=note, velocity=64, time=duration))

    midi.save(output_file)

chord_sequence = [('Cmaj7', 480), ('F7', 200), ('G7', 480)]
style_patterns = {
    'Cmaj7': [60, 64, 67, 71],  # C, E, G, B
    'F7': [65, 69, 72, 76],     # F, A, C, E
    'G7': [67, 71, 74, 77]      # G, B, D, F
}

generate_midi(chord_sequence, style_patterns, 'output.mid')


import subprocess

def synthesize_audio(midi_file, soundfont, output_wav):
    subprocess.run(['fluidsynth', '-ni', soundfont, midi_file, '-F', output_wav, '-r', '44100'])


midi_file = 'output.mid'
soundfont = '/Users/thorwhalen/Dropbox/Media/soundfonts/Arachno SoundFont - Version 1.0.sf2'
output_wav = 'output.wav'

synthesize_audio(midi_file, soundfont, output_wav)
````

# Lunar impact crater data

````python
# From 
"""
data2midi-part1.py: Maps lunar impact crater data (.csv) into musical notes (.mid).

Written by Matt Russo
www.astromattrusso.com
www.system-sounds.com
"""

#### install libraries (with pip) if necesssary################################
# import sys
# import subprocess
# for package in ['pandas','matplotlib', 'audiolazy', 'midiutil']:
#     subprocess.check_call([sys.executable, '-m', 'pip', 'install', package])

################################################
import pandas as pd   # https://pypi.org/project/pandas/
import matplotlib.pylab as plt  # https://pypi.org/project/matplotlib/
from audiolazy import str2midi # https://pypi.org/project/audiolazy/
from midiutil import MIDIFile # https://midiutil.readthedocs.io/en/1.2.1/

##############################################################################
filename = 'lunarCraterAges'  #filename of csv with data

duration_beats = 52.8 #desired duration in beats (1 beat = 1 second if bpm=60)
bpm = 60 #tempo (beats per minute)

y_scale = 0.5  #scaling parameter for y-axis data (1 = linear)

#note set for mapping (or use a few octaves of a specific scale)
note_names = ['C1','C2','G2',
             'C3','E3','G3','A3','B3',
             'D4','E4','G4','A4','B4',
             'D5','E5','G5','A5','B5',
             'D6','E6','F#6','G6','A6']

vel_min,vel_max = 35,127   #minimum and maximum note velocity

##############################################################################
def map_value(value,min_value,max_value,min_result,max_result):
    '''maps value (or array of values) from one range to another'''
    result = min_result + (value - min_value)/(max_value-min_value)*(max_result - min_result)
    return result
##############################################################################

## Load data

# This doesn't work because the file was referenced assuming locality
# df = pd.read_csv('./data/' +filename + '.csv')  #load data as a pandas dataframe

# ... so instead we'll do it this way -----------------------------
from graze import graze
import io

url = 'https://raw.githubusercontent.com/SYSTEMSounds/sonification-tutorials/main/data/lunarCraterAges.csv'
_bytes = graze(url)
df = pd.read_csv(io.BytesIO(_bytes))
# -----------------------------


n_impacts = len(df)

ages = df['age'].values    #this is a numpy array (not a list), you can do mathematical operations directly on the object
diameters = df['diameter'].values

## Compress Time
t_data = map_value(ages, min(ages), max(ages), duration_beats, 0) #compress time from Myrs to beats, largest age (oldest crater) mapped to beat 0

## Calculate duration in seconds
duration_sec = max(t_data)*60/bpm #duration in seconds (actually, onset of last note)
print('Duration:',duration_sec,'seconds')

## Normalize and scale y-axis data
y_data = map_value(diameters, min(diameters), max(diameters), 0, 1) #normalize data, so it runs from 0 to 1 (makes scaling easier)
y_data = y_data**y_scale

## Make list of MIDI numbers of chosen notes for mapping
note_midis = [str2midi(n) for n in note_names] #make a list of midi note numbers
n_notes = len(note_midis)
print('Resolution:',n_notes,'notes')

## Map y-axis data to MIDI notes and velocity
midi_data = []
vel_data = []
for i in range(n_impacts):
    note_index = round(map_value(y_data[i], 0, 1, n_notes-1, 0)) #bigger craters are mapped to lower notes
    midi_data.append(note_midis[note_index])

    note_velocity = round(map_value(y_data[i], 0, 1, vel_min, vel_max)) #bigger craters will be louder
    vel_data.append(note_velocity)

## Save MIDI file
my_midi_file = MIDIFile(1) #one track
my_midi_file.addTempo(track=0, time=0, tempo=bpm)

for i in range(n_impacts):
    my_midi_file.addNote(track=0, channel=0, pitch=midi_data[i], time=t_data[i] , duration=2, volume=vel_data[i])

with open(filename + '.mid', "wb") as f:
    my_midi_file.writeFile(f)
print('Saved',filename + '.mid')
````
